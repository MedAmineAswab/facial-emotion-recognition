```python
# pandas est une bibliothèque utilisée pour :
  # Manipuler les tableaux(DataFrame)
  # Analyser les données
  # Charger des fichiers CSV, Excel, etc.
import pandas as pd

# Lire le fichier CSV à partir du chemin spécifié
# Le convertir en DataFrame(Structure Tabulaire) composée de :
  # Lignes = des images
  # Colonnes = features(emotion, pixels, usage)
df = pd.read_csv("/content/drive/MyDrive/emotion-detection/data/raw/fer2013.csv")

# Afficher les 5 premières lignes
df.head()
```

```python

df_emotions = (
    df["emotion"] # Sélectionner la colonne emotion
    .value_counts() # Compter les occurrences par classe
    .sort_index()  # Ordonner les émotions en ordre logique (0-6)
    .rename_axis("emotion") # Nommer l'index
    .reset_index() # Transformer le résultat en DataFrame propre
)

df_emotions
```

```python
# Mapping des labels numériques vers leurs émotions correspondantes
emotion_map = {
    0 : "angry" ,
    1 : "disgust" ,
    2 : "fear" ,
    3 : "happy" ,
    4 : "sad" ,
    5 : "surprise" ,
    6 : "neutral"
}
# Remplacement des labels numériques par leurs émotions
df_emotions["emotion"] = df_emotions["emotion"].map(emotion_map)

df_emotions
```

```python
# matplotlib.pyplot est une bibliothèque utilisée pour :
 # Visualisation des données sous forme de graphiques(bar plots, histogrammes, etc.)
import matplotlib.pyplot as plt

# Histogramme des émotions
plt.bar(df_emotions['emotion'] , df_emotions['count'])

# Titre du graphique
plt.title("Répartition des émotions dans le dataset")

# Axes
plt.xlabel("Émomotions")
plt.ylabel("Nombre d'images")

# Rotation des labels
plt.xticks(rotation=45)

# Affichage
plt.show()
```
