Lorsqu’on crée une structure de dossiers dans un notebook sans utiliser Google Drive, cette structure est associée à la session en cours.
Elle est stockée dans l’environnement temporaire (/content/) et n’est donc pas permanente : elle peut disparaître lorsque la session est interrompue ou redémarrée.

En revanche, en connectant le notebook à Google Drive (drive.mount()), on peut utiliser un espace de stockage persistant. Les fichiers et dossiers créés dans /content/drive/MyDrive/ sont sauvegardés de manière permanente et restent accessibles depuis n’importe quel notebook.

Ainsi, une structure créée dans Google Drive est durable, contrairement à celle créée uniquement dans la session locale de Colab.



```python
# os est un module Python qui permet de manipuler le système de fichiers (créer des répertoires, gérer les chemins,etc.)
import os

# Connexion à Google Drive pour accéder à un stockage persistant
from google.colab import drive
drive.mount('/content/drive')

# Chemin racine du projet (stocké dans Google Drive pour éviter la perte de données)
base_path = "/content/drive/MyDrive/emotion-detection"

# La liste des répertoires à créer

folders = ["data/raw" , # Données brutes(images originales)

           "data/processed" , # Données néttoyées / transformées

           "src" , # Code principal (chargement des données, entrainement, evaluation)

           "notebooks" , # Explorer les données, tests rapides, valider des idées et visualisations

           "models" , # Modèle sauvegardé

           "app"] # Application( webcam , interface )

for folder in folders :
  # Combiner les chemins proprement
  # Empêcher l'erreur si le dossier existe déjà
  # Crée tout le chemin, même s'il y a plusieurs niveaux
  os.makedirs(os.path.join(base_path , folder) , exist_ok=True)

print("Structure créée")
```
